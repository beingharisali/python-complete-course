```python
import numpy as np

# study hours
hours = np.array([1, 2, 3, 4, 5])
end_result = np.array([0, 0, 0, 1, 1])


# initialize parameters
w = 0.0 # how strongly study hours affect passing
b = 0.0 # minimum tendency to pass without studying

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict_probability(hours, w, b):
    z = w*hours+b
    return sigmoid(z)

def loss_function(end_result, model_prediction):
    return -np.mean(end_result * np.log(model_prediction + 1e-9)+ (1-end_result)+np.log(1-model_prediction+1e-9))

learning_rate = 0.1
echoes = 3000

for i in range(echoes):
    model_prediction = predict_probability(hours, w, b)
    dw = np.mean((model_prediction - end_result)*hours)
    db = np.mean(model_prediction-end_result)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    if i % 500 == 0:
        loss = loss_function(end_result, model_prediction)
        print("Echoes", i, "loss", loss)

def predict(hours, w, b):
    probs = predict_probability(hours, w, b)
    return (probs >= 0.5).astype(int)

study_hours = 5

prob = predict_probability(study_hours, w, b)
result = predict(study_hours,w, b)

print("Probability of passing", prob)
print("Result", "Pass" if result==1 else "Fail")

```
