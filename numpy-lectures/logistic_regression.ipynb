```python
import numpy as np
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# prediction function
def predict_pass_fail(study_hours):
    score = study_hours - 8
    probability = sigmoid(score)

    if probability>=0.5:
        status = "Pass"
    else:
        status = "Fail"
    return probability, status

students = [9, 3.5, 4, 5, 7, 8]

for hours in students:
    result = predict_pass_fail(hours)

    probability = result[0]
    status = result[1]

    print("study hours", hours)
    print("Probability of passing", round(probability, 2))
    print("Result", status)
    print("-" * 10)

```
