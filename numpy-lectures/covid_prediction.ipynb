```python
import numpy as np

# study temperature
temperature = np.array([
    [98.6, 0, 98],
    [101.2, 1, 94],
    [102.5, 1, 92],     
    [99.1, 0, 97],
    [103.0, 1, 90]           
])

end_result = np.array([0, 1, 1, 0, 1])


# initialize parameters
w = np.zeros(temperature.shape[1]) # how strongly study temperature affect passing
b = 0.0 # minimum tendency to pass without studying

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict_probability(temperature, w, b):
    z = np.dot(temperature)
    return sigmoid(z)

def loss_function(end_result, model_prediction):
    return -np.mean(end_result * np.log(model_prediction + 1e-9)+ (1-end_result)+np.log(1-model_prediction+1e-9))

recovering_rate = 0.1
echoes = 3000

for i in range(echoes):
    model_prediction = predict_probability(temperature, w, b)
    dw = np.mean((model_prediction - end_result)*temperature)
    db = np.mean(model_prediction-end_result)

    w = w - recovering_rate * dw
    b = b - recovering_rate * db

    if i % 500 == 0:
        loss = loss_function(end_result, model_prediction)
        print("Echoes", i, "loss", loss)

def predict(temperature, w, b):
    probs = predict_probability(temperature, w, b)
    return (probs >= 0.5).astype(int)

study_temperature = 5

prob = predict_probability(study_temperature, w, b)
result = predict(study_temperature,w, b)

print("Probability of passing", prob)
print("Result", "Pass" if result==1 else "Fail")

```
