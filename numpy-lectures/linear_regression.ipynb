```python
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5])
y = np.array([35, 50, 65, 70, 90])
plt.bar(x, y)
# y = mx+b
multiplication = x*y
print("multiplication of x and y", multiplication)
meanOfXY = np.mean(multiplication)
print("mean of multiplication value of XY", meanOfXY)
meanOfX = np.mean(x)
print("mean of x", meanOfX)
meanOfY = np.mean(y)
print("mean of y", meanOfY)
newValOfMeans = meanOfX * meanOfY
print("new value of means", newValOfMeans)
newValOfNumerator = meanOfXY - newValOfMeans
print("new value of numerator", newValOfNumerator)
meanOfXSquare = np.mean(x**2)
print("mean of x square", meanOfXSquare)
squareOfXMean = meanOfX**2
print("square of x mean", squareOfXMean)
newValOfDenominator = meanOfXSquare - squareOfXMean
print("new value of denominator", newValOfDenominator)
m = newValOfNumerator / newValOfDenominator
print("slope", m)

intercept = meanOfY - m*meanOfX
print("intercept", intercept)

predictedValue = m*8 + intercept
print("predictedValue", predictedValue)
```
