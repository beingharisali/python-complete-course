```python
import matplotlib.pyplot as plt

x = ['Python', 'C', 'C++', 'Java', 'JavaScript', 'Assembly']
y = [90, 60, 70, 80, 90, 40]
plt.xlabel('languages', fontsize=10)
plt.ylabel('percentage popularity', fontsize=10)
plt.title('Languages Popularity Stats', fontsize=20)
colors = ['r', 'y', 'y', 'g', 'r', 'b']
plt.bar(x, y, color=colors, width=0.3)
plt.show()
```
